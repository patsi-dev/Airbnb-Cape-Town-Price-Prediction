## 3.1 Data Cleaning

**Critical Missing Values Identified:**
- `cohost_ids` & `cohost_names`: 66.7% missing → **Marked for removal**
- `professional_management`: 42.0% missing → **Requires strategic imputation**
- `bedrooms`: 17.3% missing → **Critical feature, must impute**
- `guests`: 14.3% missing → **Important capacity feature**

**Minor Missing Values (<10%):**
- Rating columns (3.7% missing)
- Fee columns (0.3%-6.3% missing)
- Other operational features (0.7%-6.7% missing)

**Target Variable Selection**
- **Primary Target**: `ttm_avg_rate_native` (12-month average rate)
- **Completeness**: 100% - no missing values
- **Alternative**: `l90d_avg_rate_native` (90-day average rate)
- **Rationale**: Stable long-term metric with complete data

````python
# Dropping columns with excessive missing values
columns_to_drop = ['cohost_ids', 'cohost_names']
df_clean = df_clean.drop(columns=columns_to_drop)
print(f"Dropped columns: {columns_to_drop}")
print(f"Dataset shape after dropping columns: {df_clean.shape}")
````

**Columns Dropped:**
- `cohost_ids` - Excessive missingness (66.7%)
- `cohost_names` - Excessive missingness (66.7%)
- **Result**: Dataset reduced to 60 columns

````python
# Imputing critical features

# bedrooms - imputing with median
bedrooms_median = df_clean['bedrooms'].median()
df_clean['bedrooms'] = df_clean['bedrooms'].fillna(bedrooms_median)
print(f"  - bedrooms: {df_clean['bedrooms'].isnull().sum()} missing remaining")

# guests - imputing with median
guests_median = df_clean['guests'].median()
df_clean['guests'] = df_clean['guests'].fillna(guests_median)
print(f"  - guests: {df_clean['guests'].isnull().sum()} missing remaining")

# professional_management - creating unknown category
df_clean['professional_management'] = df_clean['professional_management'].fillna('unknown')
print(f"  - professional_management: {df_clean['professional_management'].isnull().sum()} missing remaining")
````

**Numerical Features:**
- `bedrooms`: 17.3% missing → Imputed with median
- `guests`: 14.3% missing → Imputed with median


**Categorical Features:**
- `professional_management`: 42% missing → Created "unknown" category

````python
# Imputing minor missing values

# Rating columns - imputing with median
rating_columns = ['rating_value', 'rating_location', 'rating_cleanliness', 
                  'rating_checkin', 'rating_accuracy', 'rating_overall', 'rating_communication']

for col in rating_columns:
    if col in df_clean.columns:
        median_val = df_clean[col].median()
        df_clean[col] = df_clean[col].fillna(median_val)
        print(f"  - {col}: {df_clean[col].isnull().sum()} missing remaining")

# Fee columns - imputing with 0
fee_columns = ['cleaning_fee', 'extra_guest_fee']
for col in fee_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna(0)
        print(f"  - {col}: {df_clean[col].isnull().sum()} missing remaining")

# Other minor missing values
minor_columns = ['instant_book', 'min_nights', 'beds', 'registration']
for col in minor_columns:
    if col in df_clean.columns:
        if df_clean[col].dtype == 'object':
            # For categorical, use mode
            mode_val = df_clean[col].mode()[0] if not df_clean[col].mode().empty else 'unknown'
            df_clean[col] = df_clean[col].fillna(mode_val)
        else:
            # For numerical, use median
            median_val = df_clean[col].median()
            df_clean[col] = df_clean[col].fillna(median_val)
        print(f"  - {col}: {df_clean[col].isnull().sum()} missing remaining")

df_clean['host_name'] = df_clean['host_name'].fillna('Unknown_Host')
````

**Numerical Features:**
- `beds`: 0.7% missing → Imputed with median
- `min_nights`: 0.7% missing → Imputed with median

**Categorical Features:**
- `registration`: 0.3% missing → Imputed with mode

**Fee Columns (Zero Imputation):**
- `cleaning_fee`: 0.3% missing → Imputed with 0
- `extra_guest_fee`: 6.3% missing → Imputed with 0

**Rating Columns (Median Imputation):**
- All rating columns (3.7% missing) → Imputed with median values

````python
# Verifing no missing values remain
remaining_missing = df_clean.isnull().sum()
remaining_missing = remaining_missing[remaining_missing > 0]

if remaining_missing.empty:
    print("SUCCESS: No missing values remaining!")
else:
    print("Columns with remaining missing values:")
    for col, count in remaining_missing.items():
        print(f"  - {col}: {count} missing")

print(f"\nFinal dataset shape: {df_clean.shape}")
print(f"Total missing values: {df_clean.isnull().sum().sum()}")
````

**Final Data Cleanliness**
- **Result**: 0 missing values across all 300 rows × 60 columns

## 3.2. Feature Engineering: Nairobi-specific Features

### 3.2.1. Security Features from Amenities

**Created from Amenities Analysis:**
- `has_security_features`: Count of security amenities per listing
- `security_score`: Categorical (low/medium/high) based on security count
- **Method**: Parsed `amenities` column for security-related keywords

````python
# Step 1: Creating Security Features from Amenities

# Defining security-related amenities (Nairobi context)
security_amenities = [
    'security', 'guard', 'gated', 'cctv', 'camera', 'alarm', 
    'secure parking', 'fence', 'safety', 'watchman'
]

# Creating security flags
df_clean['has_security_features'] = 0

for amenity in security_amenities:
    df_clean['has_security_features'] += df_clean['amenities'].str.contains(
        amenity, case=False, na=False
    ).astype(int)

print(f"   - Security features detected in {df_clean['has_security_features'].sum()} listings")

# Creating security score (categorical: low, medium, high)
df_clean['security_score'] = pd.cut(
    df_clean['has_security_features'],
    bins=[-1, 0, 1, 10],
    labels=['low', 'medium', 'high']
)
print(f'\n {df_clean['security_score'].value_counts()}')
````

````python
# Defining Nairobi CBD coordinates
nairobi_cbd = (-1.2921, 36.8219)

# Calculate distance to CBD (simplified - we'll use Euclidean for now)
def calculate_distance(lat, lon, ref_point):
    """Calculate approximate distance to reference point"""
    return np.sqrt((lat - ref_point[0])**2 + (lon - ref_point[1])**2)

df_clean['distance_to_cbd'] = df_clean.apply(
    lambda row: calculate_distance(row['latitude'], row['longitude'], nairobi_cbd), 
    axis=1
)

print(f"   - Distance to CBD range: {df_clean['distance_to_cbd'].min():.3f} to {df_clean['distance_to_cbd'].max():.3f}")
````

````python
# Creating Premium Neighborhood Flags from Coordinates
# Defining approximate coordinate boundaries for premium neighborhoods in Nairobi
# Note: These are approximate boundaries and may need adjustment
neighborhood_boundaries = {
    #  Core Premium 
    'Westlands':      {'lat_min': -1.275, 'lat_max': -1.248, 'lon_min': 36.790, 'lon_max': 36.825},
    'Kilimani':       {'lat_min': -1.314, 'lat_max': -1.278, 'lon_min': 36.770, 'lon_max': 36.805},
    'Lavington':      {'lat_min': -1.292, 'lat_max': -1.250, 'lon_min': 36.740, 'lon_max': 36.790},
    'Karen':          {'lat_min': -1.375, 'lat_max': -1.300, 'lon_min': 36.650, 'lon_max': 36.760},
    'Kileleshwa':     {'lat_min': -1.295, 'lat_max': -1.260, 'lon_min': 36.770, 'lon_max': 36.805},

    # Ultra-prime / diplomatic belt 
    'Muthaiga':       {'lat_min': -1.260, 'lat_max': -1.230, 'lon_min': 36.820, 'lon_max': 36.860},
    'Runda':          {'lat_min': -1.240, 'lat_max': -1.190, 'lon_min': 36.800, 'lon_max': 36.870},
    'Gigiri':         {'lat_min': -1.245, 'lat_max': -1.215, 'lon_min': 36.790, 'lon_max': 36.830},
    'Kitisuru':       {'lat_min': -1.255, 'lat_max': -1.205, 'lon_min': 36.740, 'lon_max': 36.790},
    'Nyari':          {'lat_min': -1.240, 'lat_max': -1.210, 'lon_min': 36.770, 'lon_max': 36.800},
    'Rosslyn':        {'lat_min': -1.225, 'lat_max': -1.195, 'lon_min': 36.770, 'lon_max': 36.805},
    'Thigiri':        {'lat_min': -1.252, 'lat_max': -1.232, 'lon_min': 36.783, 'lon_max': 36.808},

    # Prime belt around Westlands / Riverside
    'Riverside':      {'lat_min': -1.285, 'lat_max': -1.265, 'lon_min': 36.790, 'lon_max': 36.815},
    'Spring Valley':  {'lat_min': -1.275, 'lat_max': -1.245, 'lon_min': 36.760, 'lon_max': 36.795},
    'Parklands':      {'lat_min': -1.280, 'lat_max': -1.245, 'lon_min': 36.802, 'lon_max': 36.835},
    'Lower Kabete':   {'lat_min': -1.255, 'lat_max': -1.225, 'lon_min': 36.720, 'lon_max': 36.760},

    # Premium pockets further out
    'Hardy':          {'lat_min': -1.370, 'lat_max': -1.335, 'lon_min': 36.670, 'lon_max': 36.715},
    'Ridgeways':      {'lat_min': -1.240, 'lat_max': -1.205, 'lon_min': 36.835, 'lon_max': 36.875},
}

# Initialize neighborhood features
df_clean['premium_neighborhood_name'] = 'Other'
df_clean['neighborhood_tier'] = 'Non-Premium'

# Define neighborhood tiers based on your classification
neighborhood_tiers = {
    'Tier 1 - Ultra Prime': ['Muthaiga', 'Runda', 'Gigiri', 'Kitisuru', 'Nyari', 'Rosslyn', 'Thigiri'],
    'Tier 2 - Core Prime': ['Westlands', 'Kilimani', 'Lavington', 'Karen', 'Kileleshwa'],
    'Tier 3 - Prime Belt': ['Riverside', 'Spring Valley', 'Parklands', 'Lower Kabete'],
    'Tier 4 - Premium Pockets': ['Hardy', 'Ridgeways']
}

# Assign neighborhoods and tiers
neighborhood_counts = {}

for neighborhood, bounds in neighborhood_boundaries.items():
    mask = (
        (df_clean['latitude'] >= bounds['lat_min']) & 
        (df_clean['latitude'] <= bounds['lat_max']) & 
        (df_clean['longitude'] >= bounds['lon_min']) & 
        (df_clean['longitude'] <= bounds['lon_max'])
    )
    
    count = mask.sum()
    neighborhood_counts[neighborhood] = count
    
    # Assign neighborhood name
    df_clean.loc[mask, 'premium_neighborhood_name'] = neighborhood

    # Assign tier
    for tier, neighborhoods in neighborhood_tiers.items():
        if neighborhood in neighborhoods:
            df_clean.loc[mask, 'neighborhood_tier'] = tier
            break

for neighborhood, count in sorted(neighborhood_counts.items(), key=lambda x: x[1], reverse=True):
    if count > 0:
        print(f"   - {neighborhood}: {count} listings")

print(f"NEIGHBORHOOD TIER DISTRIBUTION:")
print(df_clean['neighborhood_tier'].value_counts())
````

### 4.1.2. Feature Alignment Between SA and Nairobi

Since the SA model was trained on different features, we need to align the Nairobi features to the SA model's expected features.
We have a mapping defined, but note that the SA model is a pipeline that includes preprocessing and the model.
We must ensure that the Nairobi data is transformed in the same way as the SA data was during training

````python
# Defining the expected features from the SA model based on capetown notebook
sa_expected_features = [
    'bedrooms', 'beds', 'bathrooms', 'accommodates', 'host_is_superhost',
    'neighbourhood_cleansed', 'latitude', 'longitude', 'property_type',
    'room_type', 'host_listings_count', 'host_acceptance_rate',
    'review_scores_rating', 'review_scores_location', 'instant_bookable',
    'hosting_years', 'amenities_count'
]

# Map Nairobi features to SA expected features
feature_mapping = {
    'bedrooms': 'bedrooms',
    'beds': 'beds', 
    'baths': 'bathrooms',
    'guests': 'accommodates',
    'superhost': 'host_is_superhost',
    'premium_neighborhood_name': 'neighbourhood_cleansed',
    'latitude': 'latitude',
    'longitude': 'longitude', 
    'listing_type': 'property_type',
    'room_type': 'room_type',
    'instant_book': 'instant_bookable',
    'amenities_count': 'amenities_count',
    # For missing SA features, we'll need to create placeholders
}

print("Feature mapping defined:")
available_mappings = [f for f in feature_mapping.keys() if f in X.columns]
missing_mappings = [f for f in feature_mapping.keys() if f not in X.columns]

print(f"   Available mappings: {len(available_mappings)}/{len(feature_mapping)}")
print(f"   Missing in Nairobi: {missing_mappings}")
````

````python
# Check which SA features we're missing and create placeholders
missing_sa_features = [f for f in sa_expected_features if f not in feature_mapping.values()]
print(f"   Missing SA features to handle: {missing_sa_features}")
````

````python
# Create aligned datasets for SA model prediction
def create_sa_aligned_dataset(X_data, feature_mapping, sa_expected_features):
    """Create dataset aligned with SA model feature expectations"""
    aligned_data = X_data.copy()
    
    # Apply feature mappings
    for nairobi_feat, sa_feat in feature_mapping.items():
        if nairobi_feat in aligned_data.columns:
            aligned_data[sa_feat] = aligned_data[nairobi_feat]
    
    # Add placeholder for missing SA features
    for sa_feat in sa_expected_features:
        if sa_feat not in aligned_data.columns:
            if sa_feat in ['host_listings_count', 'host_acceptance_rate', 'hosting_years']:
                aligned_data[sa_feat] = 1  # Default value
            elif sa_feat in ['review_scores_rating', 'review_scores_location']:
                aligned_data[sa_feat] = aligned_data['rating_overall'] if 'rating_overall' in aligned_data.columns else 4.5
            else:
                aligned_data[sa_feat] = 0  # Default for other missing features
    
    # Ensure correct order of features
    aligned_data = aligned_data[sa_expected_features]
    
    return aligned_data
````

````python
print(f"Creating SA-aligned datasets...")
X_train_sa = create_sa_aligned_dataset(X_train, feature_mapping, sa_expected_features)
X_val_sa = create_sa_aligned_dataset(X_val, feature_mapping, sa_expected_features) 
X_test_sa = create_sa_aligned_dataset(X_test, feature_mapping, sa_expected_features)

print(f"   SA-aligned feature shapes:")
print(f"   - X_train_sa: {X_train_sa.shape}")
print(f"   - X_val_sa: {X_val_sa.shape}")
print(f"   - X_test_sa: {X_test_sa.shape}")
````

### 4.1.3. Generating SA Model Predictions

````python
print("🔧 RECREATING SA PREPROCESSOR FEATURE SET")
print("="*50)

# Based on your SA notebook preprocessor, we need the ORIGINAL feature names
# Let's define the exact numerical and categorical features from your SA notebook

# From your SA notebook, these should be the features:
sa_numerical_features = [
    'bedrooms', 'beds', 'bathrooms', 'accommodates', 'host_listings_count',
    'host_acceptance_rate', 'review_scores_rating', 'review_scores_location',
    'hosting_years', 'amenities_count', 'has_pool', 'has_hot_tub', 
    'has_ocean_view', 'has_bbq_grill', 'has_wifi', 'has_parking'
]

sa_categorical_features = [
    'host_is_superhost', 'neighbourhood_cleansed', 'property_type', 
    'room_type', 'instant_bookable'
]

sa_original_features = sa_numerical_features + sa_categorical_features

print(f"🎯 SA Model Original Features: {len(sa_original_features)}")
print(f"   Numerical: {len(sa_numerical_features)} features")
print(f"   Categorical: {len(sa_categorical_features)} features")
print(f"   All features: {sa_original_features}")

# Create dataset with EXACT original features
def create_original_sa_dataset(X_data, sa_original_features):
    """Create dataset with EXACT original features expected by SA preprocessor"""
    original_data = pd.DataFrame(index=X_data.index)
    
    # Mapping from Nairobi to SA original features
    mapping_dict = {
        'bedrooms': 'bedrooms',
        'beds': 'beds',
        'baths': 'bathrooms', 
        'guests': 'accommodates',
        'superhost': 'host_is_superhost',
        'premium_neighborhood_name': 'neighbourhood_cleansed',
        'latitude': 'latitude',
        'longitude': 'longitude',
        'listing_type': 'property_type',
        'room_type': 'room_type', 
        'instant_book': 'instant_bookable',
        'amenities_count': 'amenities_count',
        'has_pool': 'has_pool',
        'has_hot_tub': 'has_hot_tub',
        'has_ocean_view': 'has_ocean_view',
        'has_bbq_grill': 'has_bbq_grill'
    }
    
    # Apply mappings for features that exist
    for nairobi_feat, sa_feat in mapping_dict.items():
        if nairobi_feat in X_data.columns:
            original_data[sa_feat] = X_data[nairobi_feat]
    
    # Add missing features with appropriate defaults
    for feature in sa_original_features:
        if feature not in original_data.columns:
            if feature in ['has_wifi', 'has_parking']:
                # For these amenities, assume True (common in Nairobi)
                original_data[feature] = 1
            elif feature in ['host_listings_count', 'host_acceptance_rate', 'hosting_years']:
                original_data[feature] = 1  # Default host metrics
            elif feature in ['review_scores_rating', 'review_scores_location']:
                original_data[feature] = 4.5  # Default rating
            elif feature in sa_categorical_features:
                # For categorical, use first category or appropriate default
                if feature == 'host_is_superhost':
                    original_data[feature] = False
                else:
                    original_data[feature] = 'unknown'
            else:
                original_data[feature] = 0
    
    # Ensure correct order and only include expected features
    original_data = original_data[sa_original_features]
    
    return original_data

print("\n🔄 Creating ORIGINAL SA datasets...")
X_train_original = create_original_sa_dataset(X_train_fixed, sa_original_features)
X_val_original = create_original_sa_dataset(X_val_fixed, sa_original_features)
X_test_original = create_original_sa_dataset(X_test_fixed, sa_original_features)

print(f"   Original SA dataset shapes:")
print(f"   - X_train_original: {X_train_original.shape}")
print(f"   - X_val_original: {X_val_original.shape}")
print(f"   - X_test_original: {X_test_original.shape}")

# Verify we have ALL expected features
missing_in_original = [f for f in sa_original_features if f not in X_train_original.columns]
print(f"   Missing in original dataset: {missing_in_original}")

# Check data types match what preprocessor expects
print(f"\n📊 Data types in original SA dataset:")
print(X_train_original.dtypes.value_counts())
````

````python
print("🔧 QUICK FIX: ADDING MISSING COORDINATES")
print("="*50)

# Add latitude and longitude to the SA original features list
sa_original_features.extend(['latitude', 'longitude'])
sa_numerical_features.extend(['latitude', 'longitude'])  # They are numerical features

print(f"✅ Updated SA Original Features: {len(sa_original_features)}")
print(f"   Added: latitude, longitude")

# Update the datasets with the new features
def update_with_coordinates(original_data, X_data):
    """Update the original SA dataset with latitude and longitude"""
    updated_data = original_data.copy()
    
    # Add latitude and longitude from the original Nairobi data
    if 'latitude' in X_data.columns and 'longitude' in X_data.columns:
        updated_data['latitude'] = X_data['latitude']
        updated_data['longitude'] = X_data['longitude']
    else:
        # Fallback: use approximate Nairobi coordinates
        updated_data['latitude'] = -1.2921  # Nairobi CBD approx
        updated_data['longitude'] = 36.8219
    
    return updated_data

print("\n🔄 Updating datasets with coordinates...")
X_train_original_final = update_with_coordinates(X_train_original, X_train_fixed)
X_val_original_final = update_with_coordinates(X_val_original, X_val_fixed)
X_test_original_final = update_with_coordinates(X_test_original, X_test_fixed)

print(f"   Final dataset shapes:")
print(f"   - X_train_original_final: {X_train_original_final.shape}")
print(f"   - X_val_original_final: {X_val_original_final.shape}")
print(f"   - X_test_original_final: {X_test_original_final.shape}")

# Verify all features are present
missing_final = [f for f in sa_original_features if f not in X_train_original_final.columns]
print(f"   Still missing: {missing_final}")
````

````python
print("\n🔧 FIXING DATA TYPE ISSUES")
print("="*50)

def fix_data_types(X_data):
    """Fix data types to ensure compatibility with SA model"""
    X_fixed = X_data.copy()
    
    # Ensure numerical columns are numeric
    numerical_columns = [
        'bedrooms', 'beds', 'bathrooms', 'accommodates', 'host_listings_count',
        'host_acceptance_rate', 'review_scores_rating', 'review_scores_location',
        'hosting_years', 'amenities_count', 'has_pool', 'has_hot_tub', 
        'has_ocean_view', 'has_bbq_grill', 'has_wifi', 'has_parking', 'latitude', 'longitude'
    ]
    
    for col in numerical_columns:
        if col in X_fixed.columns:
            # Convert to numeric, coercing errors to NaN then fill with 0
            X_fixed[col] = pd.to_numeric(X_fixed[col], errors='coerce').fillna(0)
    
    # Ensure categorical columns are strings (not mixed types)
    categorical_columns = [
        'host_is_superhost', 'neighbourhood_cleansed', 'property_type', 
        'room_type', 'instant_bookable'
    ]
    
    for col in categorical_columns:
        if col in X_fixed.columns:
            X_fixed[col] = X_fixed[col].astype(str)
    
    return X_fixed

print("🔄 Fixing data types in all datasets...")
X_train_fixed_dtypes = fix_data_types(X_train_inferred)
X_val_fixed_dtypes = fix_data_types(X_val_inferred)
X_test_fixed_dtypes = fix_data_types(X_test_inferred)

print("✅ Data types after fixing:")
print(X_train_fixed_dtypes.dtypes.value_counts())

# Verify no missing values
print(f"\n📊 Missing values after fixing:")
missing_after_fix = X_train_fixed_dtypes.isnull().sum().sum()
print(f"   Total missing values: {missing_after_fix}")

# Final verification before prediction
print(f"\n🎯 FINAL DATA VERIFICATION:")
print(f"   Shape: {X_train_fixed_dtypes.shape}")
print(f"   Data types:")
for dtype in X_train_fixed_dtypes.dtypes.unique():
    cols = X_train_fixed_dtypes.select_dtypes(include=[dtype]).columns.tolist()
    print(f"     {dtype}: {len(cols)} columns")
````

````python
print("\n🎯 FINAL ATTEMPT WITH FIXED DATA TYPES")
print("="*50)

if final_model is not None:
    try:
        print("   Generating predictions with FIXED data types...")
        sa_train_pred = final_model.predict(X_train_fixed_dtypes)
        sa_val_pred = final_model.predict(X_val_fixed_dtypes)
        sa_test_pred = final_model.predict(X_test_fixed_dtypes)
        
        print("🎉 SUCCESS! SA model predictions generated!")
        print(f"   Training set: {sa_train_pred.shape}")
        print(f"   Validation set: {sa_val_pred.shape}") 
        print(f"   Test set: {sa_test_pred.shape}")
        
        # Add to our datasets
        X_train_transfer = X_train.copy()
        X_val_transfer = X_val.copy()
        X_test_transfer = X_test.copy()
        
        X_train_transfer['sa_model_prediction'] = sa_train_pred
        X_val_transfer['sa_model_prediction'] = sa_val_pred
        X_test_transfer['sa_model_prediction'] = sa_test_pred
        
        transfer_success = True
        
        # Analyze the predictions
        print("\n📈 ANALYZING SA PREDICTIONS:")
        sa_train_pred_original = np.expm1(sa_train_pred)
        train_prices_original = np.expm1(y_train)
        
        correlation = np.corrcoef(sa_train_pred_original, train_prices_original)[0,1]
        mape = np.mean(np.abs((sa_train_pred_original - train_prices_original) / train_prices_original)) * 100
        
        print(f"   Correlation with Nairobi prices: {correlation:.3f}")
        print(f"   MAPE on Nairobi data: {mape:.1f}%")
        
        # Quick visualization
        plt.figure(figsize=(10, 4))
        
        plt.subplot(1, 2, 1)
        plt.scatter(sa_train_pred_original, train_prices_original, alpha=0.6)
        max_val = max(sa_train_pred_original.max(), train_prices_original.max())
        plt.plot([0, max_val], [0, max_val], 'r--', alpha=0.8)
        plt.xlabel('SA Predictions (KES)')
        plt.ylabel('Actual Nairobi Prices (KES)')
        plt.title(f'Correlation: {correlation:.3f}')
        plt.grid(True, alpha=0.3)
        
        plt.subplot(1, 2, 2)
        error_pct = (sa_train_pred_original - train_prices_original) / train_prices_original * 100
        plt.hist(error_pct, bins=30, alpha=0.7, color='orange')
        plt.axvline(0, color='red', linestyle='--', linewidth=2)
        plt.xlabel('Prediction Error (%)')
        plt.ylabel('Frequency')
        plt.title(f'MAPE: {mape:.1f}%')
        plt.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        # Business interpretation
        if correlation > 0.5:
            print("🎯 EXCELLENT: SA model works well with Nairobi data!")
        elif correlation > 0.3:
            print("📊 GOOD: SA model provides meaningful signals")
        else:
            print("⚠️  MODERATE: SA model has limited transfer value")
            
    except Exception as e:
        print(f"❌ Final attempt with fixed data types failed: {e}")
        print("   This suggests fundamental incompatibility between datasets")
        print("   Using business-aware placeholder")
        transfer_success = False
else:
    transfer_success = False

if not transfer_success:
    print("🔄 Using business-aware placeholder for transfer learning")
    # Ensure placeholder is set up
    X_train_transfer = X_train.copy()
    X_val_transfer = X_val.copy()
    X_test_transfer = X_test.copy()
    
    # Use the sophisticated business-aware placeholder
    def business_aware_placeholder(row):
        base_price = 2000  # Base price per bedroom
        
        # Bedroom premium
        bedrooms = row.get('bedrooms', 1)
        bedroom_component = bedrooms * base_price
        
        # Neighborhood tier premium (from our actual data)
        tier_premiums = {
            'Tier 1 - Ultra Prime': 6000,
            'Tier 2 - Core Prime': 3000, 
            'Tier 3 - Prime Belt': 1500,
            'Tier 4 - Premium Pockets': 750,
            'Non-Premium': 0
        }
        tier = row.get('neighborhood_tier', 'Non-Premium')
        tier_component = tier_premiums.get(tier, 0)
        
        # Property type premium
        property_type = row.get('property_type_group', 'entire_property')
        if property_type == 'entire_property':
            type_component = 1000
        elif property_type == 'private_room':
            type_component = 0
        else:
            type_component = 500
            
        # Security premium (Nairobi-specific)
        security = row.get('security_score', 'low')
        if security == 'high':
            security_component = 800
        elif security == 'medium':
            security_component = 400
        else:
            security_component = 0
            
        total_price = bedroom_component + tier_component + type_component + security_component
        return np.log1p(total_price)  # Convert to log space
    
    X_train_transfer['sa_model_prediction'] = X_train_transfer.apply(business_aware_placeholder, axis=1)
    X_val_transfer['sa_model_prediction'] = X_val_transfer.apply(business_aware_placeholder, axis=1)
    X_test_transfer['sa_model_prediction'] = X_test_transfer.apply(business_aware_placeholder, axis=1)
````
